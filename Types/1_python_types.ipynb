## Type Hints

Python has support for optional "type hints" (also called "type annotations").

These **"type hints"** or annotations are a special syntax that allows declaring the type of a variable.

By declaring types for your variables, editors and tools can give you better support.

```python
'''Just specify the type of function parameters and return type''' 

def get_full_name(first_name: str, last_name: str) -> str: 
    return first_name + " " + last_name

print(get_full_name("Deepak", "Krishna"))
```

### Simple Types

Standard Types in Python

- `int`
- `float`
- `bool`
- `str`
- `bytes`

```python
def get_items(a: str, b: int, c: float, d: bool, e: bytes): 
    return a, b, c, d, e
```

### `typing` Module

the typing module has some additional types that can be used for special use cases

```python
from typing import Any

def some_function(data: Any) -> Any: 
    return data
```

### Generic Types

Some types can take "type parameters" in square brackets, to define their internal types. 

These types that can take type parameters are called Generic types or Generics

- `list`
- `tuple`
- `set`
- `dict`

```python
def process_item(items: list[str]): 
    for item in items: 
        print(item)
```

```python
# you can do the same with tuple and set
def process_items(t: tuple[int, int, str], s: set[bytes]): 
    return t, s
```

This means:

- The variable `items_t` is a `tuple` with 3 items, an `int`, another `int`, and a `str`.
- The variable `items_s` is a `set`, and each of its items is of type `bytes`.

```python
# same goes for dict too
def process_items(prices: dict[str, float]): 
    for name, price in prices.items(): 
        print(name, price)
```

## Union

You can declare a variable with several types. To define it, use the vertical bar '|' to separate the types

This is called a "union", because the variable can be anything in the union of those two sets of types

```python
def process_items(item: int | str): 
    return item
```

### None

You can declare that a value could have a type, like `str`, but that it oculd also be `None`

```python
def say_hi(name: str | None = None): 
    if name is not None: 
        print(f"Hey {name}")
    else: 
        print("Hello, World!")
```
