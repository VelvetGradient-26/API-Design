## Classes and Objects

A class is just a template. When you create an object from a class, it follows that template


```python
class Student: 
    def __init__(self, name: str, age: int, gender: str):
        self.name = name
        self.age = age
        self.gender = gender

    def __str__(self):
        result = f"Name: {self.name}\nAge: {self.age}\nGender: {self.gender}"
        return result
        
# Instanciation
student = Student("Deepak", 20, "Male")
print(student)
```

## Inheritance

Inheritance allows one class to acquire the properties and methods of another class. This avoids writing the same code again and again. 

- Parent class (base class) contains common, shared features
- Child Class (derived class) inherits everything from the parent
- Child class can also add its won new attributes and methods on top
- Changes made in the parent class automatically reflect in child class

```python
class Animal: 
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    def eat(self): 
        print("I am eating")

    def sleep(self): 
        print("I am sleeping")

class Dog(Animal): 
    def __init__(self, name, age, breed):
        super().__init__(name, age)
        self.breed = breed
    def bark(self): 
        print("I am barking")

dog = Dog("Tom", 3, "German Shepherd")
dog.bark()
dog.eat()
dog.sleep()

print(dog.name, dog.age, dog.breed)
```

## Polymorphism

Poly mean many, morph means forms. 

Polymorpism allows the same method name to behave differently depending on which object calls it.This makes your code highly flexible and extensible

- Same method name works differently for different classes
- You don't need to remember multiple method names for similar actions
- Method Overriding: Child class provides its own version of a parent's method

Instead of writing fragile `if/elif` chains to check an object's type before acting on it, polymorphism lets you treat different objects uniformly through a shared interface. 

Python implements polymorphism in four primary ways

### 1. Duck Typing (The most Pythonic approach)

In Python, an object's suitability is determined by the methods and properties it has, not is class inheritance ("If it walks like a duck and quacks like a duck, then it must be a duck")

```python
class CreditCard: 
    def pay(self, amount): 
        print(f"Charing {amount} to credit card.")

class PayPal: 
    def pay(self, amount): 
        print(f"Sending ${amount} via PayPal")

class CryptoWallet: 
    def pay(self, amount): 
        print(f"Transferring ${amount} in BTC")


# Polymorphic function: doesn't care about the class, only that the .pay() method exists
def process_checkout(payment_method, total): 
    payment_method.pay(total)

# All three work seemlessly despite having no shared parent class
process_checkout(CreditCard(), 50)
process_checkout(PayPal(), 75)
process_checkout(CryptoWallet(), 700.4)
```

### 2. Inheritance & Method Overriding

This is the classic Object-Oriented approach. A base class defines a default method, and subclasses override it to provide specialized behaviour

```python
class Notification: 
    def send(self, message): 
        raise NotImplementedError("Subclasses must implement this method")

class EmailNotification(Notification): 
    def send(self, message): 
        print(f"[EMAIL] Sending: {message}")

class SMSNotification(Notification): 
    def send(self, message): 
        print(f"[SMS] Texting: {message}")

alerts = [EmailNotification(), SMSNotification()]

for alert in alerts: 
    alert.send("Your server CPU is at 95%!!!")
```

### 3. Operator Overloading (Dunder Methods)

Python's built-in operators (+, *, ==, len()) are polymorphic, you can modify them at your conveniece

```python
class Vector: 
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return F"Vector({self.x, self.y})"


v1 = Vector(2, 3)
v2 = Vector(4, 1)

print(v1 + v2)
```

### 4. Protocols (Static Duck Typing)

As codebases grow, pure duck typing can make it hard for IDEs and type checkers to catch erros before runtime. typing.Protocol gives you the safety of interfaces without forcing classes to inherit from a base class

```python
from typing import Protocol

class Exportable(Protocol): 
    def export(self) -> str: 
        ...

class PDFReport: 
    def export(self) -> str: 
        return "Exporting data as PDF..."

class CSVReport: 
    def export(self) -> str: 
        return "Exporting data as CSV..."

# Type-safe polymorphism without inheritance
def download_file(item: Exportable) -> None: 
    print(item.export())

download_file(PDFReport())
download_file(CSVReport())
```

## Encapsulation

Encapsulation is about protecting the internal data of a class. You control what the outside world can see and what it cannot. This prevents accidental modification of important data. 

- Data and methods that operate on that data are bundled together inside a class
- Sensitive data is marked private so it can't be accessed directly from outside
- Access is given through controlled methods called **getters** and **setters** 
- Makes code more secure and predictable

Example: ATM machine you can check balance and withdraw, but you can't access the bank's internal system directly

```python
class Bank: 
    def __init__(self, name: str, balance: int): 
        self.name: str = name
        # Private variable (double underscore)
        self.__balance: int = balance

    @property # property decorator allows you to use methods like attributes 
    def balance(self):
        return self.__balance

    def deposit(self, amount: float): 
        self.__balance += amount
        print(f"Amount Deposited\nCurrent Balance: {self.__balance}\n")

    def withdraw(self, amount: float): 
        if amount > self.__balance: 
            print("Not Enough balance in account")
        else: 
            self.__balance -= amount
            print(f"Amount withdrawn\nCurrent Balance: {self.__balance}\n")

account = Bank("Deepak", 10_000)
account.deposit(25_000)
account.withdraw(5_000)

print(account.balance)
```

## Abstraction

Abstraction is about reducing complexity. You expose only the necessary details to the user and hide all the complicated internal logic. The user doesn't need to know how things work inside, just that they work

```python
from abc import ABC, abstractmethod

# An abstract class, the class must inherited and all the methods inside it must be implemented
class Shape(ABC): 
    def area(self): 
        pass

    def perimeter(self): 
        pass

class Rectangle(Shape): 
    def __init__(self, length: float, breadth: float): 
        self.length: float = length
        self.breadth: float = breadth

    @abstractmethod
    def area(self): 
        return self.length * self.breadth

    @abstractmethod
    def perimeter(self): 
        return 2 * (self.length + self.breadth)
```
