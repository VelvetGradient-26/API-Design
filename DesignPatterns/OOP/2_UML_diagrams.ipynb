# Unified Language Modelling

UML is a standard way to visually design and represent software systems using diagrams.

The main purpose is to have a common visual language that shows how classes, objects, relationships and interactions work in a system

This makes it easier for developers and teamsn to understand, discuss and plan the system before writing actual code

## Advantages of using UML

- **Visualization**: UML diagrams let you see the system visually, like a blueprint. This makes it much easier to understand how different parts of the system are structured, how they connect and how they work together. 

- **Documentation**: UML diagrams act as detailed documentation of your software design. This is very helpful when you need to maintain the code later or when the system needs to grow bigger. 

- **Communication**: UML diagrams create a common language that designers, developers and teams can all understand. Everyone can look at the same diagram and discuss the design clearly withou confusion

- **Standardization**: Since UML is a universally accepted standard, everyone in the software industry reads and understands these diagrams the same way. No matter where you work, UML will be the same.

## Basic Elements

1. **Class**: A blueprint or template for cr3eating objects. It has a name, properties (attributes) and actions (methods) that objects can perform. 

```python
class User: 
    def __init__(self, name, age): 
        self.name = name
        self.age = age
    
    def login(self): 
        pass
```

2. **Interface**: In Python, we use Abstract Classes to define which methods a class must have. It tells what a class should do, not how to do it.

```python
from abc import ABC, abstractmethod

class Loginable(ABC): 
    @abstractmethod
    def login(self): 
        pass

```

3. **Object**: A real working copy of a class created when the program runs. If Class is a blueprint, Object is the actual house built from it. 

4. **Association**: A connect between two classes showing how their objects interact or relate to each other.

5. **Inheritance**: An `is-a` relationship where a child class inherites properties and methods from a parent class. 

6. **Composition**: A strong relationship where one object is made up of another object and cannot exist without it
Example: A Car has an Engine, if the car is destroyed, the engine is destroyed too

7. **Aggregation**: A weaker realtionship where one object contains another, but both can exist independently. 
Example: A Classroom has Students, if the classroom closes, students still exist
